```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[6,7,8,9,10]
plt.plot(x,y)
plt.xlabel("X axis")
plt.ylabel('Y Axis')
plt.title('Simple graph')
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[6,7,8,9,10]
plt.plot(x,y)
plt.plot(y,x)
plt.xlabel("X axis")
plt.ylabel('Y Axis')
plt.title('Simple graph')
plt.show()
```

```python
import numpy as np
x=np.linspace(0,10,100)
y=np.sin(x)
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.xlabel("X axis")
plt.ylabel('Y axis sinx values')
plt.title('Sin Wave')
plt.grid(True)
plt.show()
```
